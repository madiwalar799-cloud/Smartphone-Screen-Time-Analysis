```python
plt.figure(figsize=(8,5))

plt.hist(df['Screen On Time (hours/day)'],
         bins=10,
         edgecolor='black')

plt.title('Distribution of Smartphone Screen Time')
plt.xlabel('Screen On Time (Hours per Day)')
plt.ylabel('Number of Users')

plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(data=df, x='Age', bins=10, kde=True)

plt.title('Age Distribution of Smartphone Users')
plt.xlabel('Age')
plt.ylabel('Number of Users')

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.barplot(data=df, x='Gender',
            y='Screen On Time (hours/day)',
            errorbar=None)

plt.title('Average Screen Time by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Screen Time (Hours)')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(data=df,
            x='Operating System',
            y='Screen On Time (hours/day)',
            errorbar=None)

plt.title('Screen Time by Operating System')
plt.xlabel('Operating System')
plt.ylabel('Average Screen Time (Hours)')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(data=df,
             x='App Usage Time (min/day)',
             bins=15,
             kde=True)

plt.title('Daily App Usage Time Distribution')
plt.xlabel('App Usage Time (Minutes per Day)')
plt.ylabel('Number of Users')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=df,
                x='Screen On Time (hours/day)',
                y='Battery Drain (mAh/day)')

plt.title('Screen Time vs Battery Drain')
plt.xlabel('Screen On Time (Hours per Day)')
plt.ylabel('Battery Drain (mAh/day)')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df,
              x='User Behavior Class')

plt.title('Distribution of User Behavior Classes')
plt.xlabel('User Behavior Class')
plt.ylabel('Number of Users')

plt.show()
```

```python
plt.figure(figsize=(10,7))

corr = df.select_dtypes(include='number').drop(
    columns=['User ID']
).corr()

sns.heatmap(corr,
            annot=True,
            cmap='coolwarm',
            fmt='.2f')

plt.title('Correlation Heatmap of Smartphone Usage')

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../dataset/user_behavior_dataset.csv")

print("Dataset loaded successfully!")
```

```python
print(df.shape)
```

```python
print("Average Screen Time:")
print(df['Screen On Time (hours/day)'].mean())

print("\nAverage App Usage Time:")
print(df['App Usage Time (min/day)'].mean())

print("\nAverage Battery Drain:")
print(df['Battery Drain (mAh/day)'].mean())
```

```python
gender_analysis = df.groupby('Gender')['Screen On Time (hours/day)'].mean()

print(gender_analysis)
```

```python
os_analysis = df.groupby('Operating System')['Screen On Time (hours/day)'].mean()

print(os_analysis)
```

```python
behavior_analysis = df['User Behavior Class'].value_counts().sort_index()

print(behavior_analysis)
```
